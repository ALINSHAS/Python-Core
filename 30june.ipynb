```python
#for else
for i in range(1,5):
    print(i)
else:
    print("hello")
```

```python
#while else
i=1
while(i<=5):
    print(i)
    i=i+1
else:
    print("hello")
```

```python
#for break else
for i in range(1,5):
    if(i==5):
        break
    print(i)
        
else:#will execute only after completing the iteration
    print("hello")
```

```python
#write a program to check whether a number is prime or not
n=int(input("Enter a number"))
if(n>1):
    for i in range(2,n):
        if(n%i==0):
            print("number is not  prime")
        break
else:
     print("number is  prime")
```

```python
#check wheather a number is perfect number
n = int(input("Enter a number: "))

sum = 0

for i in range(1, n):
    if n % i == 0:
        # print(i)
        sum = sum + i

# print("Sum =", sum)

if sum == n:
    print("Perfect Number")
else:
    print("Not a Perfect Number")
```
