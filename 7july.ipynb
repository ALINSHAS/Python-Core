```python
# define a function to check whether the number is Prime or not
def prime(n):
    for i in range(2,n):
        if n % i == 0:
            print(n, "is not a prime number")
            break
    print(n, "is a prime number")
n = int(input("Enter a number: "))
prime(n)
```

```python
#define a function that takes a string and character and return the number of occurence of character in the string
def occurenece(s,ch):
    count=0
    for i in s:
        if i==ch:
            count+=1
    return count


s=input("Enter the string")
ch=input("enter the character")
result=occurenece(s,ch)
print("the number of occurence of character: ",result)
```
