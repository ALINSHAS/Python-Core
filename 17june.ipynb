```python
#interchange two variable using temo varible
a=3
b=5

#swap variable
temp=a
a=b
b=temp
print(a,b)
```

```python
#interchange two variable
a=3
b=5

#swap variable
a,b=b,a
print(a,b)
```

```python
#[] index operator = position of string

s="Hello world"
print(s[0])     #a letter in a string print
print(s[-1])    #reverse letter position print
print(len(s))   #length of a string
```

```python
#slice operator

s="Hello world"
print(s[0:5])
print(s[6:11])
print(s[0:5:2])
```

```python
#given string
s="python coding is easy and fun"

#length of the string
print(len(s))

#reverse of string
print(s[::-1])

#print the first character
print(s[0])

# print the last character
print(s[-1])

#print the characters from 2nd index to 10th character
print(s[2:11])

#print 13th character
print(s[12])

#print 5th index character
print(s[5])
```
