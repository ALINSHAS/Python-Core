```python
l = [12, 34, 56, 78, 13, 87, 5]

for i in l:
    if i > 1:
        prime = True
        for j in range(2, i):
            if i % j == 0:
                prime = False
                break
        if prime:
            print(i)
```

```python
#print all prime number in range(1,100)
for i in range(1,101):
    if i > 1:
        for j in range (2,i):
            if i % j ==0 :
                break
        else:
            print(i)
```
