```python
l = [12, 34, 56, 45, 89, 16, 69, 33]

# Print each number
for i in l:
    print(i)

# Print even numbers
print("Even numbers:")
for i in l:
    if i % 2 == 0:
        print(i)

# Print numbers divisible by 5
print("Divisible by 5:")
for i in l:
    if i % 5 == 0:
        print(i)

# Print numbers containing digit '3'
print("Contains digit 3:")
for i in l:
    if '3' in str(i):
        print(i)

# Sum of list
sum = 0
for i in l:
    sum += i
print("Sum =", sum)

# Count of numbers
count = 0
for i in l:
    count += 1
print("Count =", count)

# Count of even numbers
count = 0
for i in l:
    if i % 2 == 0:
        count += 1
print("Even Count =", count)

# Product of list
product = 1
for i in l:
    product *= i
print("Product =", product)

# print square of each number
for i in l:
    print(i * i)
```

```python
#Given a list of colours
colors = ["red", "green", "blue", "orange", "black"]
#print starting letter of each item
for i in colors:
    print(i[0])
#print last letter
for i in colors:   
    print(i[-1])
#give reverse of each colors
for i in colors:
    print(i[::-1])
#print all colors with "b"
for i in colors:
    if('b' in i):
        print(i)
    
```
