```python
#List comprehension
l=[1,2,3,4]
new=[i**2 for i in l]
print(new)
new=[5 for i in l]
print(new)
```

```python
#Given a list 
l=[12,45,34,67,90]
#create a new list with even values
new=[i for i in l if i%2==0]
print(new)
#create a new list with value greater than 50
new=[i for i in l if i > 50]
print(new)
#Given a list colors
colors=['red',"green","blue","orange"]
#create a new list with first letter of each colour
new=[i[0] for i in colors]
print(new)
#create a new list length of each color
new=[len(i) for i in colors]
print(new)
```
