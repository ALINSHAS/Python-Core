```python
l=[1,2,3,4,5,6,7,8,9,0]
print(list(filter(lambda x:x%2==0,l)))
```

```python
#given  a list
l=[23,67,90,12,45,52]
#number greater than 50
print(list(filter(lambda i:i>50 ,l)))
```

```python
f=['apple','orange',"grapes",'pineapple',"grapes"]
#fruitname whose length is greater than 5
print(list(filter(lambda i: len(i) > 5, f)))
```

```python
l=[34,78,96,23,12,15]
#number is divisible by 3 & 5
print(list(filter(lambda x:x%3==0 and x%5==0,l)))
```

```python
#Reduce
l=[1,2,3,4]
import functools
print(functools.reduce(lambda x,y:x+y,l,0))
```
