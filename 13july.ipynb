```python
square=lambda x:x**2
print(square(4))
```

```python
cube=lambda x:x**3
print(cube(3))
```

```python
length=lambda a:len(a)
print(length("hello"))
```

```python
d=lambda a:a["name"]
print(d({'name': "arun","Age": 23}))
```

```python
last=lambda a:a[0]
print(last('harry'))
```

```python
sum=lambda a,b,c,d:a+b+c+d
print(sum(4,5,6,7))
```

```python
l=[1,2,3,4]
print(map(lambda x:x**2,l))
print(list(map(lambda x:x**2,l)))
print(set(map(lambda x:x**2,l)))
print(tuple(map(lambda x:x**2,l)))
```

```python
#create a new list of cubes
l=[12,3,5,6,7]
print(list(map(lambda x:x**3,l)))
```

```python
#create a new list of square root
l=[1,2,3,4]
print(list(map(lambda x:x*0.5,l)))
```

```python
#create a new list of length
colors=["red","green","blue","yellow","black"]
print(list(map(lambda x:len(x),colors)))
#create a new list of first characters
print(list(map(lambda i:i[0],colors)))
#create a new list of last characters
print(list(map(lambda i:i[-1],colors)))
#create a new list of reverse of each elemnt
print(list(map(lambda i:i[::-1],colors)))
```

```python
#Given a list 
l=[23,78,12,56]
#Add 10 to each element in the given sequence
print(list(map(lambda i:i+10,l)))
```

```python
#given a list of dictionaries
l=[{'empid':100,'name':'arun','salary':20000,'email':'arun@gmail.com'},
   {'empid':101,'name':'amal','salary':25000,'email':'amal@gmail.com'},
   {'empid':102,'name':'anu','salary':30000,'email':'anu@gmail.com'}]

# create a new list of emails
print(list(map(lambda i:i["email"],l)))
```
