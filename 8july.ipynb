```python
#define a function that takes as arguments and return a new list containing unique elements from list
def new(l):
    new=[]
    for i in l:
        if i not in new:
            new.append(i)
    return new 

result=new([12,34,78,12,67,34,90,23])
print(result)
```

```python
#define a function that takes 2 list as argumnets and print the common elements

def common(l1,l2):
    new=[]
    for i in l1:
        if i in l2:
            new.append(i)
    return new

result = common([12, 34, 56, 78, 90], [90, 34, 11, 57, 45])
print(result)

```

```python
#keyword type
def f(n,a):
    print(n)
    print(a)
f(n="arun",a=23)
```

```python
def f(*args):
    print(args)
f(1,2,3,4,5,6,7,8,9)
f(2,4,6)
```

```python
def f(**kwargs):
    print(kwargs)
f(name="arun",age=23)
f(name="amal",age=24,place="ernakulam")
```

```python
#define a function that accept any numebr and return their sum
def sumof(*args):
    sum=0
    for i in args:
        sum=sum+i
    return sum
result=sumof(1,2,3,4,5,6,7,8)
print(result)
```
